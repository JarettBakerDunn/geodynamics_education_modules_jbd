## Description

An introduction to regression with examples as applied to geophysical problems. Regression provides a way to quantify relationships between observed geophysical data and the underlying model parameters that produced them, allowing us to make predictions about parameters we cannot observe directly. This makes it a valuable tool for parameter estimation in the Earth, where the interior is largely inaccessible to direct measurement.

### Learning objectives

Upon completion of this module you would be familiar with the following techniques:

**Machine learning**
- Linear regression
- Polynomial regression
- Random forest


### Prerequisites
- Python (> version 3.3) with the following libraries installed:
  * NumPy
  * pandas
  * sklearn
  * BurnMan
  * Matplotlib

- Fundamental data science concepts such as:
  * Training and testing datasets
  * Error metrics such as mean squared error (MSE) and R-squared ($R^2$)
  * Overfitting and underfitting

## Glossary

### Machine Learning
Machine learning is the process of using statistical methods to generate models from training data. Once a model is generated, it can accurately predict the most likely output values for inputs which are not necessarily within the training data. Machine learning techniques are also used to study and manipulate the structure of datasets themselves. [IGyTheM](https://geodynamics.org/highlight-february2026) is an example of machine learning applied to the geosciences.

### Supervised & Unsupervised Machine Learning
Different machine learning algorithms can be classified as "supervised" or "unsupervised".

In [supervised learning](https://en.wikipedia.org/wiki/Supervised_learning), machine learning algorithms rely on training datasets which have been labelled by a human. Supervised algorithms are more often used to make predictions based on input or sort inputs into categories. For example, [Linear Regression](https://en.wikipedia.org/wiki/Linear_regression) is a supervised algorithm which is used for prediction and relies on a labelled set of training data. 

A possible application might be to use construct a linear regression model to predict how likely a volcano is to erupt in the future. This is an example of supervised learning because the data inputs and outputs are labelled: we know how many times tremors were measured near the volcano in a year, and we know how many times that volcano erupted in that year. Using this labelled data you could use linear regression to predict how likely a recently discovered volcano (which lacks historical eruption data) is to erupt given that n number of tremors were measured over the last year. In the real world, it is unlikely that a simple model like linear regression would actually be a good predictor of likely eruptions. The relationship between tremor frequency and eruptions might not be linear, and there are many more variables and unknowns. However, more complex supervised learning techniques follow the same principle as linear regression. Each point of data connects an input value to a label/output value.


In [unsupervised learning](https://en.wikipedia.org/wiki/Unsupervised_learning), algorithms operate on datasets which have not been labelled by a human. Unlabelled data means that the model is not given any information beforehand about what the data might mean. Unsupervised algorithms are often used for analysing and manipulating datasets themselves. They are also used for sorting new data into categories. For example, [k-means clustering](https://en.wikipedia.org/wiki/K-means_clustering) is an unsupervised algorithm which is used to investigate how datapoints are clustered throughout their distributions.

You might use an unsupervised machine learning technique like k-means clustering to categorize types of volanoes based on eruption and tremor frequency. Here, learning is unsupervised because You don't provide the model with the types of possible volcanoes beforehand. Instead, the model identifies categories from the data it is given. Creating categories using unsupervised ML algorithms makes it possible to investigate trends or structure in the dataset and sort future inputs into categories.

There are also [semi-supervised](https://en.wikipedia.org/wiki/Weak_supervision) models. Semi-supervised models have been provided with some labelled data, but are also trained on unlabelled data. Because semi-supervised techniques lie between unsupervised and supervised learning, the exact method used to generate models can vary widely. A common pattern are self-learning semi-supervised algorithms, which are provided with a small set of labeled data and a large unlabeled dataset. During the process of training the model iteratively labels the unlabelled training data. Another common technique is to pre-train a model using unsupervised learning, and then to provide labels later. The most famous application of semi-supervised learning are probably large language models or LLMs.

### Typical Steps in Machine Learning Workflows

#### Step 1: Data cleaning
Sometimes datasets need to be [cleaned](https://www.geeksforgeeks.org/data-analysis/data-cleaning-introduction/). Cleaning a dataset is the process of removing datapoints which would cause problems when training or testing a model. This includes missing, nonsensical and outlying values. For example, values which are NaN or null, values which are extremely high or low, or values which made innaccurate by formatting issues. Cleaning data is important because any statistical model is only as accurate as the data it is trained on. If the dataset is already high quality, this step is not always needed.

#### Step 2: Splitting Datasets Into Train/Test/Validation subsets
ML models need to be evaluated using data which they have not been trained on. Otherwise, the evaluation would not accurately represent the performance of the model on real world data. Practically this means that datasets need to be [split](https://developers.google.com/machine-learning/crash-course/overfitting/dividing-datasets) into separate subsets for training, testing, and validation. The training subset is used to actually train the model. The testing dataset is used for tuning the model after it has been trained. Tuning means making adjustments to aspects of the model in order to produce a better score when scored using the testing data. Finally, the validation subset is used to evaluate the model after tuning is completed. A validation subset separate from the testing subset is needed because the model has already been exposed to the testing subset. In effect, the evaluating the model using the validation subset checks that the model has not become overfitted to the testing subset during the tuning phase.

- [pyTorch](https://docs.pytorch.org/docs/2.14/index.html)
- [Tensorflow](https://www.tensorflow.org/learn)

#### Step 4: Model Training / Tuning

Model training refers to using a machine learning algorithm to create a statistical model based on your training data. In libraries like scikit-learn this can look as simple as calling python function, although the actual algorithms being run on the data can obviously be very complex. 

Tuning is an iterative process where models are trained, evaluated on test data, and then retrained with some parameters of the ML algorithm tweaked until evaluation from the test data is good enough. Evaluation is based on quantitative error metrics which measure how much the output of the model varies from the output of the test data. 

<img src='./images/workflow_with_validation_set.svg' width="500" align="center"/>

Image: Google, [Machine Learning Crash Course](https://developers.google.com/machine-learning/crash-course/overfitting/dividing-datasets)

The specific error metric or metrics depend on the machine learning model being evaluated. For regression models like Linear Regression [Root Mean Squared Error](https://en.wikipedia.org/wiki/Root_mean_square_deviation) (RMSE) is commonly used. The RMSE is the average of the differences between the observed output values from the test dataset and the predicted outputs for all input values in the test dataset. Tuning a regression model might  involve tweaking parameters of the regression function until the RMSE is low enough.

Evaluating models which classify data rather than predict future outputs works differently. Rather than measuring the error of predicted outputs, classification models are evaluated by measuring how often they make correct predictions. 

There are four terms which are important in understanding how classification models are evaluated. 

- True Positive: The model sorts the data as being in the class it actually is in.
- False Positive: The model sorts the data as being in the class, although the data actually belongs to a different class.
- False Negative: The model sorts the data into a different class than the data actually belongs to.
- True Negative: The model sorts the data into a different class, when the truly does not belong to the class in question.

These terms apply to each class which the model can sort data into. For example, a model might sort rocks into mafic, ultramafic, or felsic categories depending on the minerals present. Here are those terms defined in the context of this example:

- True Positive: The model categorizes a rock as being felsic it is felsic.
- False Positive: The model categorizes a rock as being felsic when it is mafic or ultramafic.
- False Negative: The model categorizes a rock as being mafic or ultramafic when it felsic
- True Negative: The model categorizes a rock as being mafic or ultramafic when it is mafic or ultramafic


Now, the [metrics](https://developers.google.com/machine-learning/crash-course/classification/accuracy-precision-recall) which build on these terms:

Precision is the ratio of true positives to all predicted positives. In the felsic/mafic/ultramafic example this would measure the fraction of rocks classified as felsic which were actually felsic. This metric is useful when measuring how likely the model is to produce false positives.

Recall is the ratio of true positives to all actual positives. In the felsic/mafic/ultramafic example this would measure the proportion of felsic rocks which are correctly categorized as felsic. This metric is useful when measuring how likely the model is to correctly identify positive cases, regardless of if it also identifies false positives.

Accuracy is the ratio of correct classifications (true positive and true negative) to all classifications made. In the felsic/mafic/ultramafic example, this would mean the proportion of rocks which are correctly classified into the felsic/mafic/ultramafic categories. This is a broad metric which is not necessarily always useful for real world datasets.

#### Step 5: Model Validation
After a model scores well enough on error metrics during the tuning phase, it is time to test how well the model would perform on data which it has not yet been exposed to. The model is evaluated using the same error metrics used during the tuning phase, but on the validation dataset instead of the test dataset. This is to make sure that the model is not [overfit](https://developers.google.com/machine-learning/crash-course/overfitting/overfitting) (tuned too closely) to the testing data. When a model is overfit, it means that the model is so closely matched to the training and/or testing data that it fails at making accurate predictions on data from outside the testing/training dataset. It is also possible for a model to be underfit, but this is easier to catch since it will score poorly on error metrics when tuning the model. A model which is underfit is not sufficiently trained on its training data, so that it fails to make accurate predictions even on data which it has been trained on.

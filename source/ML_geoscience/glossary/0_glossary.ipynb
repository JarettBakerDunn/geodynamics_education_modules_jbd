## Glossary

### Machine Learning
Machine learning relies on using statistical methods to generate models from a set of training data. Once a model is generated, it can accurately predict the most likely output values for inputs which are not necessarily within the training data. [IGyTheM](https://geodynamics.org/highlight-february2026) is an example of machine learning applied to the geosciences.

### Supervised & Unsupervised Machine Learning
Different machine learning algorithms can be classified as "supervised" or "unsupervised".

In [supervised learning](https://en.wikipedia.org/wiki/Supervised_learning), machine learning algorithms rely on training datasets which have been labelled by a human. Supervised algorithms are more often used to make predictions based on input or sort inputs into categories. For example, [k-nearest neighbors (KNN)](https://en.wikipedia.org/wiki/K-nearest_neighbors_algorithm) is a supervised algorithm which is used for prediction and classification, and relies on a labelled set of training data.

In [unsupervised learning](https://en.wikipedia.org/wiki/Unsupervised_learning), algorithms operate on datasets which have not been labelled by a human. Unsupervised algorithms are more often used for analysing and manipulating datasets themselves. For example, [k-means clustering](https://en.wikipedia.org/wiki/K-means_clustering) is an unsupervised algorithm which is used to investigate how datapoints are clustered throughout their distributions.

### Overfitting
A common problem in machine learning is [overfitting](https://developers.google.com/machine-learning/crash-course/overfitting/overfitting), where a model is so closely matched to its training data that it fails at making accurate predictions on data from outside of its training set.

### Evaluating Models Using Error Metrics
Machine Learning models need to be evaluated. Models which predict values are evaluated by [examining](https://en.wikipedia.org/wiki/Coefficient_of_determination) how much real data points from the test set differ from their predicted values. The actual method used to depends on the Machine Learning algorithm used by the model. Models which classify inputs are evaluated by taking the ratio of correct categorization.

### Splitting Datasets
It is important to evaluate machine learning models using data that is new to the model. If a model scores well when being tested using the same data that it was trained on, it may be impossible to tell if the model is performing well because it is an accurate model, or because overfit to the training data.

Often when working with data only one large dataset is available.
When this is the case, the dataset is often [split](https://developers.google.com/machine-learning/crash-course/overfitting/dividing-datasets) into training and testing subsets. 

### The scikit-learn Machine Learning Library

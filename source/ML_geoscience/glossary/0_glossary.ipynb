## Glossary

### Machine Learning
Machine learning relies on using statistical methods to generate models from a set of training data. Once a model is generated, it can accurately predict the most likely output values for inputs which are not necessarily within the training data. [IGyTheM](https://geodynamics.org/highlight-february2026) is an example of machine learning applied to the geosciences.

### Supervised & Unsupervised Machine Learning
Different machine learning algorithms can be classified as "supervised" or "unsupervised".

In [supervised learning](https://en.wikipedia.org/wiki/Supervised_learning), machine learning algorithms rely on training datasets which have been labelled by a human. Supervised algorithms are more often used to make predictions based on input or sort inputs into categories. For example, [Linear Regression](https://en.wikipedia.org/wiki/Linear_regression) is a supervised algorithm which is used for prediction and relies on a labelled set of training data. 

A possible application might be to use construct a linear regression model to predict how likely a volcano is to erupt in the future. This is an example of supervised learning because the data inputs and outputs are labelled: we know how many times tremors were measured near the volcano in a year, and we know how many times that volcano erupted in that year. Using this labelled data you could use linear regression to predict how likely a recently discovered volcano (which lacks historical eruption data) is to erupt given that n number of tremors were measured over the last year. In the real world, it is unlikely that a simple model like linear regression would actually be a good predictor of likely eruptions. The relationship between tremor frequency and eruptions might not be linear, and there are many more variables and unknowns. However, more complex supervised learning techniques follow the same principle as linear regression. Each point of data connects an input value to a label/output value.


In [unsupervised learning](https://en.wikipedia.org/wiki/Unsupervised_learning), algorithms operate on datasets which have not been labelled by a human. Unlabelled data means that the model is not given any information beforehand about what the data might mean. Unsupervised algorithms are often used for analysing and manipulating datasets themselves. They are also used for sorting new data into categories. For example, [k-means clustering](https://en.wikipedia.org/wiki/K-means_clustering) is an unsupervised algorithm which is used to investigate how datapoints are clustered throughout their distributions.

You might use an unsupervised machine learning technique like k-means clustering to categorize types of volanoes based on eruption and tremor frequency. Here, learning is unsupervised because You don't provide the model with the types of possible volcanoes beforehand. Instead, the model identifies categories from the data it is given. Creating categories using unsupervised ML algorithms makes it possible to investigate trends or structure in the dataset and sort future inputs into categories.

There are also [semi-supervised](https://en.wikipedia.org/wiki/Weak_supervision) models. Semi-supervised models have been provided with some labelled data, but are also trained on unlabelled data. Because semi-supervised techniques lie between unsupervised and supervised learning, the exact method used to generate models can vary widely. A common pattern are self-learning semi-supervised algorithms, which are provided with a small set of labeled data and a large unlabeled dataset. During the process of training the model iteratively labels the unlabelled training data. Another common technique is to pre-train a model using unsupervised learning, and then to provide labels later. The most famous application of semi-supervised learning are probably large language models or LLMs.

### Overfitting
A common problem in machine learning is [overfitting](https://developers.google.com/machine-learning/crash-course/overfitting/overfitting), where a model is so closely matched to its training data that it fails at making accurate predictions on data from outside of its training set.

### Evaluating Models Using Error Metrics
Machine Learning models need to be evaluated. Models which predict values are evaluated by [examining](https://en.wikipedia.org/wiki/Coefficient_of_determination) how much real data points from the test set differ from their predicted values. The actual method used to depends on the Machine Learning algorithm used by the model. Models which classify inputs are evaluated by taking the ratio of correct categorization.

### Splitting Datasets
It is important to evaluate machine learning models using data that is new to the model. If a model scores well when being tested using the same data that it was trained on, it may be impossible to tell if the model is performing well because it is an accurate model, or because overfit to the training data.

Often when working with data only one large dataset is available.
When this is the case, the dataset is often [split](https://developers.google.com/machine-learning/crash-course/overfitting/dividing-datasets) into training and testing subsets.

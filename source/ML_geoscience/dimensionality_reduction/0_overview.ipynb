## Description

An introduction to dimensionality reduction and how it can be used in geochemical problems. Dimensionality reduction techniques help us visualize and understand complex, high-dimensional data by projecting it onto a lower-dimensional space while preserving important structures and patterns. This makes it easier to identify trends, groupings, and anomalies in the data that might not be apparent in the original high-dimensional space.

### Learning objectives

Upon completion of this module you would be familiar with the following techniques:

**Machine learning**
- Principal Component Analysis (PCA)
- t-distributed Stochastic Neighbor Embedding (t-SNE)

### Pre-requisites
- Python (> version 3.3) with the following libraries installed:
  * numPy
  * pandas
  * sklearn
  * scipy
  * matplotlib

- Fundamental geochemistry knowledge, including the concept of isotopes and isotope ratios, and the geochemical characteristics of ocean island basalts (OIBs) and mantle components.
